# Gradient Descent

```python
import torch
import torch.nn.functional as F
```

```python
# 목표 텐서 생성
target = torch.FloatTensor([[.1, .2, .3],
                            [.4, .5, .6],
                            [.7, .8, .9]])
```

```python
# 초기 x를 target과 같은 크기의 랜덤 값으로 생성
x = torch.rand_like(target)
# gradient 추적 활성화
x.requires_grad = True
```

```python
# MSE 손실 계산
loss = F.mse_loss(x, target)
```

```python
# 학습 파라미터 설정
threshold = 1e-5
learning_rate = 1.0
iter_cnt = 0

# 손실이 threshold보다 클 동안 반복
while loss > threshold:
    iter_cnt += 1
    
    # gradient 계산
    loss.backward()
    
    # gradient descent 업데이트
    x = x - learning_rate * x.grad
    
    # gradient 추적을 위한 detach 및 requires_grad 재설정
    x.detach_()
    x.requires_grad_(True)
    
    # 새로운 loss 계산
    loss = F.mse_loss(x, target)
    
    # 진행 상황 출력
    print('%d-th Loss: %.4e' % (iter_cnt, loss))
    print(x)
```
