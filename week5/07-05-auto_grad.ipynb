# PyTorch AutoGrad

```python
import torch
```

```python
# requires_grad=True로 텐서 생성
x = torch.FloatTensor([[1, 2],
                       [3, 4]]).requires_grad_(True)
```

```python
# 연산 그래프 구성
x1 = x + 2
x2 = x - 2
x3 = x1 * x2
y = x3.sum()
```

```python
# backward로 gradient 계산
y.backward()
```

```python
# x의 gradient 확인
print(x.grad)
```

```python
# detach()로 gradient 추적 해제 후 numpy 변환
print(x3.detach_().numpy())
```
