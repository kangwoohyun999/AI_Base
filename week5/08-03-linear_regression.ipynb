```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import torch
```

```python
# Boston 데이터셋 로드
from sklearn.datasets import load_boston
boston = load_boston()
```

```python
# DataFrame으로 변환
df = pd.DataFrame(boston.data, columns=boston.feature_names)
df["TARGET"] = boston.target
```

```python
# 학습 데이터 준비 (간단한 선형 회귀 예시)
# 실제 모델 학습 코드는 notebook에 따라 추가
X = torch.FloatTensor(df.drop('TARGET', axis=1).values)
y = torch.FloatTensor(df['TARGET'].values).view(-1, 1)
```

```python
# 모델 파라미터 초기화
w = torch.randn((13, 1), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
```

```python
# 학습 루프
optimizer = torch.optim.SGD([w, b], lr=0.001)
for epoch in range(100):
    optimizer.zero_grad()
    y_pred = X @ w + b
    loss = torch.mean((y_pred - y) ** 2)
    loss.backward()
    optimizer.step()
    if epoch % 20 == 0:
        print(f'Epoch {epoch}: Loss = {loss.item():.4f}')
```

```python
# 결과 시각화
y_hat = (X @ w + b).detach()
df_plot = pd.DataFrame(torch.cat([y, y_hat], dim=1).numpy(), columns=["y", "y_hat"])
sns.pairplot(df_plot, height=5)
plt.show()
```
